```python
import numpy as np
import matplotlib.pyplot as plt
```

## Parte A. Altura maxima de Betelgeuse

La formula geometrica mas directa es:

$$
h_{max} = 90^\circ - |\phi - \delta|
$$


```python
def altura_maxima(lat_grados, declinacion_grados):
    return 90.0 - abs(lat_grados - declinacion_grados)


def horas_sobre_altura(lat_grados, declinacion_grados, altura_min_grados=30.0):
    phi = np.deg2rad(lat_grados)
    delta = np.deg2rad(declinacion_grados)
    h0 = np.deg2rad(altura_min_grados)
    cos_H = (np.sin(h0) - np.sin(phi) * np.sin(delta)) / (np.cos(phi) * np.cos(delta))
    cos_H = np.clip(cos_H, -1.0, 1.0)
    H = np.rad2deg(np.arccos(cos_H))
    return 2 * H / 15.0


lat_bogota = 4.6481
lat_medellin = 6.25
dec_betelgeuse = 7.4

h_bog = altura_maxima(lat_bogota, dec_betelgeuse)
h_med = altura_maxima(lat_medellin, dec_betelgeuse)
horas_bog = horas_sobre_altura(lat_bogota, dec_betelgeuse)
horas_med = horas_sobre_altura(lat_medellin, dec_betelgeuse)

print(f"Bogota:   altura maxima = {h_bog:.4f} deg, tiempo sobre 30 deg = {horas_bog:.4f} h")
print(f"Medellin: altura maxima = {h_med:.4f} deg, tiempo sobre 30 deg = {horas_med:.4f} h")
```

**Conclusion:** la afirmacion de que Betelgeuse no supera 30 grados no cuadra con la cuenta analitica. La estrella si sube muy alto para esas latitudes.


## Parte B. Ejercicio final de repaso: segunda derivada

La formula pedida al final del notebook es:

$$
f''(x) \approx \frac{f(x+h) - 2 f(x) + f(x-h)}{h^2}
$$


```python
def second_order_diff(f, x, h):
    return (f(x + h) - 2 * f(x) + f(x - h)) / h**2


x0 = 1.0
h_values = np.array([1e-1, 5e-2, 1e-2, 5e-3, 1e-3])

exact = -np.sin(x0)
approx = second_order_diff(np.sin, x0, h_values)
error = np.abs(exact - approx)
slope = np.polyfit(np.log10(h_values), np.log10(error), 1)[0]

print("Errores:", error)
print(f"Pendiente ajustada: {slope:.6f}")
```

```python
plt.figure(figsize=(7, 4.5))
plt.loglog(h_values, error, "o-", label="error de la segunda derivada")
plt.xlabel("h")
plt.ylabel("error absoluto")
plt.title("Verificacion del orden del metodo")
plt.grid(True, which="both", ls="--", alpha=0.4)
plt.legend()
plt.show()
```

**Resultado:** la pendiente sale aproximadamente `2`, asi que el metodo es de orden `O(h^2)`.
