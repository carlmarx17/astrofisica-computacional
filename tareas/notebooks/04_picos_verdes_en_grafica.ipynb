# Solucion 4: por que aparecen los picos verdes

Este notebook rehace la idea central de `02. More Diferentiation Examples.ipynb`.
La pregunta es por que la curva verde, que corresponde a la diferencia centrada, muestra picos cuando `h` es demasiado pequeno.


```python
import numpy as np
import matplotlib.pyplot as plt


def forward_diff(f, x, h):
    return (f(x + h) - f(x)) / h


def backward_diff(f, x, h):
    return (f(x) - f(x - h)) / h


def centered_diff(f, x, h):
    return (f(x + h) - f(x - h)) / (2 * h)
```

```python
x0 = np.pi / 2
exact = np.cos(x0)
h_values = np.logspace(-16, -1, 250)

err_forward = np.abs(forward_diff(np.sin, x0, h_values) - exact)
err_backward = np.abs(backward_diff(np.sin, x0, h_values) - exact)
err_centered = np.abs(centered_diff(np.sin, x0, h_values) - exact)
```

```python
fig, axes = plt.subplots(1, 2, figsize=(12, 4.5), constrained_layout=True)

axes[0].loglog(h_values, err_forward, label="forward", color="tab:red")
axes[0].loglog(h_values, err_backward, label="backward", color="tab:blue")
axes[0].loglog(h_values, err_centered, label="centered", color="tab:green")
axes[0].set_xlabel("h")
axes[0].set_ylabel("error absoluto")
axes[0].set_title("Rango completo")
axes[0].grid(True, which="both", ls="--", alpha=0.35)
axes[0].legend()

zoom = h_values < 1e-8
axes[1].loglog(h_values[zoom], err_forward[zoom], color="tab:red")
axes[1].loglog(h_values[zoom], err_backward[zoom], color="tab:blue")
axes[1].loglog(h_values[zoom], err_centered[zoom], color="tab:green")
axes[1].set_xlabel("h")
axes[1].set_ylabel("error absoluto")
axes[1].set_title("Zoom en h muy pequeno")
axes[1].grid(True, which="both", ls="--", alpha=0.35)

plt.show()
```

## Explicacion clara

La curva verde baja mas rapido al principio porque el metodo centrado tiene error de truncamiento `O(h^2)`.

Pero cuando `h` se vuelve demasiado pequeno, pasa esto:

- se restan dos numeros casi iguales: `f(x+h)` y `f(x-h)`
- esa resta pierde cifras significativas
- el error de redondeo deja de ser pequeno comparado con la senal real
- al dividir entre `2h`, el ruido relativo se amplifica

Eso produce los picos verdes. En resumen: no es que el metodo centrado sea malo; choca con el limite de precision de maquina.
