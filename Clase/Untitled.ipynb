```python
import math

# Definir la función a maximizar
def A(theta):
    return 4 * math.sin(theta) * (1 + math.cos(theta))

# Inicialización de variables
xl = 0.0                  
xu = math.pi / 2.0        
epsilon = 0.05            

# Proporción de la sección áurea
R = (math.sqrt(5) - 1) / 2

# Paso 1 - Puntos intermedios iniciales
d = R * (xu - xl)
x1 = xl + d   
x2 = xu - d   

# Paso 2 - Evaluar función
f1 = A(x1)
f2 = A(x2)

# --- NUEVO FORMATO DE IMPRESIÓN (ENCABEZADO) ---
print("=" * 70)
print("🔍 INICIANDO BÚSQUEDA DE LA SECCIÓN ÁUREA")
print("=" * 70)
print(f"{'Iter':^6} | {'x_l':^8} | {'x_u':^8} | {'f(x_1)':^8} | {'f(x_2)':^8} | {'Error':^8}")
print("-" * 70)

iteracion = 1
# Criterio de tolerancia
while abs(xu - xl) >= epsilon:
    error_actual = abs(xu - xl)
    
    # --- NUEVO FORMATO DE IMPRESIÓN (FILA DE DATOS) ---
    # Usamos '^' para centrar el texto en las columnas
    print(f"{iteracion:^6} | {xl:^8.4f} | {xu:^8.4f} | {f1:^8.4f} | {f2:^8.4f} | {error_actual:^8.4f}")
    
    # Paso 3 - Actualizar el intervalo
    if f1 > f2:
        xl = x2
        x2 = x1
        f2 = f1
        d = R * (xu - xl)
        x1 = xl + d
        f1 = A(x1)
    else:
        xu = x1
        x1 = x2
        f1 = f2
        d = R * (xu - xl)
        x2 = xu - d
        f2 = A(x2)
        
    iteracion += 1

# Calcular el punto medio final
theta_opt = (xl + xu) / 2
area_max = A(theta_opt)

# --- NUEVO FORMATO DE IMPRESIÓN (RESULTADOS FINALES) ---
print("=" * 70)import math

# Definir la función original, primera y segunda derivada
def A(theta):
    return 4 * math.sin(theta) * (1 + math.cos(theta))

def dA(theta):
    return 4 * math.cos(theta) + 4 * math.cos(2 * theta)

def d2A(theta):
    return -4 * math.sin(theta) - 8 * math.sin(2 * theta)

# Inicialización
x_i = math.pi / 4.0      # Estimación inicial razonable (45 grados / 0.7854 rad)
epsilon = 0.05           # Criterio de tolerancia

print("=" * 75)
print("🚀 INICIANDO BÚSQUEDA CON EL MÉTODO DE NEWTON")
print("=" * 75)
print(f"{'Iter':^6} | {'x_i':^10} | {'A(x_i)':^10} | {'A''(x_i)':^10} | {'A''''(x_i)':^10}")
print("-" * 75)

iteracion = 1
while True:
    # Evaluación
    fx = A(x_i)
    dfx = dA(x_i)
    d2fx = d2A(x_i)
    
    print(f"{iteracion:^6} | {x_i:^10.4f} | {fx:^10.4f} | {dfx:^10.4f} | {d2fx:^10.4f}")
    
    # Criterio de tolerancia: Si el valor absoluto de la derivada es menor a epsilon, se detiene
    if abs(dfx) < epsilon:
        break
        
    # Iteración: Fórmula del Método de Newton para optimización
    x_next = x_i - (dfx / d2fx)
    
    # Actualizar para el siguiente ciclo
    x_i = x_next
    iteracion += 1

print("=" * 75)
print("🏆 RESULTADOS FINALES DE LA OPTIMIZACIÓN")
print("-" * 75)
print(f" ▸ Ángulo óptimo encontrado (θ*) : {x_i:.4f} radianes")
print(f" ▸ Área máxima calculada A(θ*)   : {A(x_i):.4f}")
print(f" ▸ Iteraciones totales           : {iteracion}")
print(f" ▸ Derivada final A'(θ*)         : {abs(dfx):.5f} (Meta: < {epsilon})")
print("=" * 75)print("RESULTADOS FINALES DE LA OPTIMIZACIÓN")
print("-" * 70)
print(f" ▸ Ángulo óptimo encontrado (θ*) : {theta_opt:.4f} radianes")
print(f" ▸ Área máxima calculada A(θ*)   : {area_max:.4f}")
print(f" ▸ Iteraciones totales           : {iteracion - 1}")
print(f" ▸ Tamaño final del intervalo    : {abs(xu-xl):.4f} (Meta: < {epsilon})")
print("=" * 70)
```

```python
import math

# Definir la función de dos variables a maximizar
def f(l, theta):
    return (18 - 2*l + 2*l*math.cos(theta)) * l * math.sin(theta)

# Tolerancia y Proporción Áurea global
epsilon = 0.05
R = (math.sqrt(5) - 1) / 2

# --- CICLO INTERNO: Encuentra el mejor theta para un 'l' fijo ---
def max_theta_para_l(l_fijo):
    xl_t = 0.0
    xu_t = math.pi / 2.0
    dt = R * (xu_t - xl_t)
    t1 = xl_t + dt
    t2 = xu_t - dt
    
    f1 = f(l_fijo, t1)
    f2 = f(l_fijo, t2)
    
    # Búsqueda Áurea estándar para theta
    while abs(xu_t - xl_t) >= epsilon:
        if f1 > f2:
            xl_t = t2
            t2 = t1
            f2 = f1
            dt = R * (xu_t - xl_t)
            t1 = xl_t + dt
            f1 = f(l_fijo, t1)
        else:
            xu_t = t1
            t1 = t2
            f1 = f2
            dt = R * (xu_t - xl_t)
            t2 = xu_t - dt
            f2 = f(l_fijo, t2)
            
    mejor_theta = (xl_t + xu_t) / 2
    area_max_local = f(l_fijo, mejor_theta)
    return area_max_local, mejor_theta

# --- CICLO EXTERNO: Encuentra el mejor 'l' ---
print("=" * 70)
print("🎯 INICIANDO BÚSQUEDA DE SECCIÓN ÁUREA ANIDADA (2D)")
print("=" * 70)
print(f"{'Iter':^6} | {'l_evaluado':^12} | {'Mejor_theta':^12} | {'Área_Max':^12} | {'Error (l)':^10}")
print("-" * 70)

xl_l = 0.0
xu_l = 9.0
dl = R * (xu_l - xl_l)
l1 = xl_l + dl
l2 = xu_l - dl

# Evaluamos los primeros puntos pidiendo al ciclo interno que nos dé el área máxima
area1, theta1_opt = max_theta_para_l(l1)
area2, theta2_opt = max_theta_para_l(l2)

iteracion = 1
# Búsqueda Áurea estándar para la longitud l
while abs(xu_l - xl_l) >= epsilon:
    error_l = abs(xu_l - xl_l)
    print(f"{iteracion:^6} | {l1:^12.4f} | {theta1_opt:^12.4f} | {area1:^12.4f} | {error_l:^10.4f}")
    
    if area1 > area2:
        xl_l = l2
        l2 = l1
        area2 = area1
        dl = R * (xu_l - xl_l)
        l1 = xl_l + dl
        area1, theta1_opt = max_theta_para_l(l1)
    else:
        xu_l = l1
        l1 = l2
        area1 = area2
        dl = R * (xu_l - xl_l)
        l2 = xu_l - dl
        area2, theta2_opt = max_theta_para_l(l2)
        
    iteracion += 1

# Calcular resultados finales
l_optimo = (xl_l + xu_l) / 2
area_final, theta_optimo = max_theta_para_l(l_optimo)

print("=" * 70)
print(" RESULTADOS FINALES DE LA SECCIÓN ÁURA PURA")
print("-" * 70)
print(f" ▸ Longitud óptima (l*) : {l_optimo:.4f} cm")
print(f" ▸ Ángulo óptimo (θ*)   : {theta_optimo:.4f} radianes")
print(f" ▸ Base resultante (b)  : {18 - 2*l_optimo:.4f} cm")
print(f" ▸ Área máxima A(l,θ)   : {area_final:.4f} cm²")
print("=" * 70)
```
