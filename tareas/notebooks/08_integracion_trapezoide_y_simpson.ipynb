# Solución 8: Integración — Trapezoide y Simpson

Origen: `01. Fundamental Algorithms/03. Integration/Integracion.pdf`

La integral que representa la distancia recorrida por el cohete entre $t=8$ y $t=30$ segundos:

$$x = \int_{8}^{30} \left[2000\ln\!\left(\frac{140000}{140000-2100t}\right) - 9.8t\right] dt$$

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy import integrate

# ============================================================
# FUNCIÓN Y VALOR EXACTO DE REFERENCIA
# ============================================================
def v(t):
    """Velocidad ascendente del cohete [m/s]."""
    return 2000 * np.log(140000 / (140000 - 2100*t)) - 9.8*t

a, b = 8.0, 30.0
x_exacto, _ = integrate.quad(v, a, b)
print(f"Valor exacto (scipy): {x_exacto:.6f} m")
print()
```

---
## Parte 1: Regla del Trapezoide simple

**Ejercicio para entregar:** Use la regla del trapezoide para estimar la distancia cubierta entre $t=8$ y $t=30$ s.

$$I \approx \frac{b-a}{2}\left[f(a) + f(b)\right]$$

```python
# ============================================================
# REGLA DEL TRAPEZOIDE SIMPLE
# ============================================================
def trapezoide_simple(f, a, b):
    """Regla del trapezoide con un solo intervalo."""
    return (b - a) / 2 * (f(a) + f(b))

T_simple = trapezoide_simple(v, a, b)
err_abs = abs(T_simple - x_exacto)
err_rel = err_abs / abs(x_exacto) * 100

print("=" * 55)
print("REGLA DEL TRAPEZOIDE SIMPLE")
print("=" * 55)
print(f"  v(t=8)  = {v(a):.4f} m/s")
print(f"  v(t=30) = {v(b):.4f} m/s")
print(f"  Aproximación     : {T_simple:.4f} m")
print(f"  Valor exacto     : {x_exacto:.4f} m")
print(f"  Error absoluto   : {err_abs:.4f} m")
print(f"  Error relativo   : {err_rel:.4f} %")
print("=" * 55)
```

---
## Parte 2: Regla del Trapezoide compuesto

**Ejercicio para entregar:** Use la regla compuesta del trapezoide con $n=2$ segmentos.

$$I \approx \frac{h}{2}\left[f(x_0) + 2\sum_{i=1}^{n-1}f(x_i) + f(x_n)\right], \quad h = \frac{b-a}{n}$$

```python
# ============================================================
# REGLA DEL TRAPEZOIDE COMPUESTO
# ============================================================
def trapezoide_compuesto(f, a, b, n):
    """
    Regla del trapezoide compuesto con n subintervalos.
    """
    h = (b - a) / n
    xi = np.linspace(a, b, n + 1)   # n+1 puntos
    fi = f(xi)
    return h / 2 * (fi[0] + 2 * np.sum(fi[1:-1]) + fi[-1])


print("=" * 58)
print("REGLA DEL TRAPEZOIDE COMPUESTO — cohete")
print("=" * 58)
for n in [2, 4, 8, 16]:
    T = trapezoide_compuesto(v, a, b, n)
    err_abs = abs(T - x_exacto)
    err_rel = err_abs / abs(x_exacto) * 100
    print(f"  n={n:>2}: {T:.4f} m  |  err_abs={err_abs:.4f}  |  err_rel={err_rel:.4f}%")
print("=" * 58)
print()

# Detalle n=2 (ejercicio)
n2 = 2
h2 = (b - a) / n2
xi2 = np.linspace(a, b, n2 + 1)
print(f"Detalle n=2: h={h2}  puntos={xi2}")
print(f"  f(t8)={v(xi2[0]):.4f}, f(t19)={v(xi2[1]):.4f}, f(t30)={v(xi2[2]):.4f}")
T2 = trapezoide_compuesto(v, a, b, 2)
print(f"  Trapezoide compuesto (n=2): {T2:.4f} m")
print(f"  Error absoluto: {abs(T2 - x_exacto):.4f} m")
```

## Parte 3: Regla de Simpson 1/3

**Ejercicio para entregar:** Use la regla de Simpson 1/3 con $n=2$ segmentos (3 puntos).

$$I \approx \frac{h}{3}\left[f(x_0) + 4f(x_1) + f(x_2)\right], \quad h = \frac{b-a}{2}$$

```python
# ============================================================
# REGLA DE SIMPSON 1/3 (simple y compuesto)
# ============================================================
def simpson_1_3_simple(f, a, b):
    """Simpson 1/3 simple (3 puntos: a, (a+b)/2, b)."""
    h = (b - a) / 2
    m = (a + b) / 2
    return h / 3 * (f(a) + 4*f(m) + f(b))


def simpson_1_3_compuesto(f, a, b, n):
    """
    Simpson 1/3 compuesto con n subintervalos (n debe ser par).
    """
    if n % 2 != 0:
        raise ValueError("n debe ser par para Simpson 1/3 compuesto")
    h = (b - a) / n
    xi = np.linspace(a, b, n + 1)
    fi = f(xi)
    return h / 3 * (fi[0] + 4*np.sum(fi[1:-1:2]) + 2*np.sum(fi[2:-2:2]) + fi[-1])


S13 = simpson_1_3_simple(v, a, b)
err_abs = abs(S13 - x_exacto)
err_rel = err_abs / abs(x_exacto) * 100

print("=" * 55)
print("REGLA DE SIMPSON 1/3 SIMPLE — cohete")
print("=" * 55)
m_val = (a + b) / 2
h_val = (b - a) / 2
print(f"  t0={a}  t1={m_val}  t2={b}")
print(f"  v(t0)={v(a):.4f}  v(t1)={v(m_val):.4f}  v(t2)={v(b):.4f}")
print(f"  h = {h_val}")
print(f"  Aproximación   : {S13:.4f} m")
print(f"  Valor exacto   : {x_exacto:.4f} m")
print(f"  Error absoluto : {err_abs:.4f} m")
print(f"  Error relativo : {err_rel:.4f} %")
print("=" * 55)

print()
print("Convergencia Simpson 1/3 compuesto:")
for n in [2, 4, 8, 16]:
    S = simpson_1_3_compuesto(v, a, b, n)
    ea = abs(S - x_exacto)
    er = ea / abs(x_exacto) * 100
    print(f"  n={n:>2}: {S:.4f} m  |  err_abs={ea:.6f}  |  err_rel={er:.6f}%")
```

## Parte 4: Regla de Simpson 3/8 — función de elección propia

La regla de Simpson 3/8 usa **4 puntos** (3 subintervalos) y tiene orden $O(h^4)$:

$$I \approx \frac{3h}{8}\left[f(x_0) + 3f(x_1) + 3f(x_2) + f(x_3)\right], \quad h = \frac{b-a}{3}$$

**Función elegida:** $f(x) = \sin(x)e^{-0.2x}$, $x \in [0, \pi]$

Valor exacto (por partes): $\int_0^\pi \sin(x)e^{-0.2x}\,dx$

```python
# ============================================================
# REGLA DE SIMPSON 3/8 — función propia
# ============================================================
def simpson_3_8_simple(f, a, b):
    """Simpson 3/8 simple (4 puntos, 3 subintervalos)."""
    h = (b - a) / 3
    x0, x1, x2, x3 = a, a+h, a+2*h, b
    return 3*h/8 * (f(x0) + 3*f(x1) + 3*f(x2) + f(x3))


def simpson_3_8_compuesto(f, a, b, n):
    """
    Simpson 3/8 compuesto con n subintervalos (n múltiplo de 3).
    """
    if n % 3 != 0:
        raise ValueError("n debe ser múltiplo de 3 para Simpson 3/8")
    h = (b - a) / n
    xi = np.linspace(a, b, n + 1)
    fi = f(xi)
    result = fi[0] + fi[-1]
    for i in range(1, n):
        if i % 3 == 0:
            result += 2 * fi[i]
        else:
            result += 3 * fi[i]
    return 3*h/8 * result


# Función elegida
g = lambda x: np.sin(x) * np.exp(-0.2*x)
ag, bg = 0.0, np.pi

g_exacto, _ = integrate.quad(g, ag, bg)
S38_simple  = simpson_3_8_simple(g, ag, bg)

err_abs_38 = abs(S38_simple - g_exacto)
err_rel_38 = err_abs_38 / abs(g_exacto) * 100

print("=" * 60)
print("REGLA DE SIMPSON 3/8 — f(x) = sin(x)·e^(-0.2x), x∈[0,π]")
print("=" * 60)
h38 = (bg - ag) / 3
pts = [ag, ag+h38, ag+2*h38, bg]
print("  Puntos: " + "  ".join([f"x{i}={p:.4f}" for i, p in enumerate(pts)]))
print("  Valores: " + "  ".join([f"f(x{i})={g(p):.4f}" for i, p in enumerate(pts)]))
print(f"  h = {h38:.4f}")
print(f"  Simpson 3/8 (simple)  : {S38_simple:.6f}")
print(f"  Valor exacto (scipy)  : {g_exacto:.6f}")
print(f"  Error absoluto        : {err_abs_38:.6f}")
print(f"  Error relativo        : {err_rel_38:.4f} %")
print("=" * 60)

print()
print("Convergencia Simpson 3/8 compuesto:")
for n in [3, 6, 12, 24]:
    S = simpson_3_8_compuesto(g, ag, bg, n)
    ea = abs(S - g_exacto)
    er = ea / abs(g_exacto) * 100
    print(f"  n={n:>2}: {S:.8f}  |  err_abs={ea:.8f}  |  err_rel={er:.6f}%")
```

```python
# Visualización comparativa de todos los métodos — cohete
metodos = ['Trapezoide\nsimple', 'Trapezoide\ncompuesto (n=2)', 'Simpson 1/3\nsimple', 'Exacto']
valores  = [
    trapezoide_simple(v, a, b),
    trapezoide_compuesto(v, a, b, 2),
    simpson_1_3_simple(v, a, b),
    x_exacto
]
colores = ['#e74c3c', '#f39c12', '#27ae60', '#2c3e50']

fig, ax = plt.subplots(figsize=(9, 5))
bars = ax.bar(metodos, valores, color=colores, width=0.5, edgecolor='black', linewidth=0.7)
ax.axhline(x_exacto, color='black', linestyle='--', linewidth=1.5, label=f'Exacto = {x_exacto:.2f} m')

for bar, val in zip(bars, valores):
    ax.text(bar.get_x() + bar.get_width()/2, bar.get_height() + 50,
            f'{val:.1f}', ha='center', va='bottom', fontsize=9)

ax.set_ylabel('Distancia [m]')
ax.set_title('Comparación de métodos de integración — Cohete', fontsize=13)
ax.legend()
ax.set_ylim(min(valores)*0.85, max(valores)*1.08)
plt.tight_layout()
plt.show()
```
