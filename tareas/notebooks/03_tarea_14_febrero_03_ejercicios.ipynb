Este notebook deja los tres ejercicios en formato de entrega corta:

1. aceleracion de un pulsar
2. gradiente de presion estelar
3. derivadas de un corrimiento al rojo


```python
import numpy as np
```

```python
# 1) Pulsar
h = 0.5
velocidades = np.array([200.0, 205.2, 210.8, 216.9, 223.5])

respuesta_1 = {
    "forward_en_t0": (velocidades[1] - velocidades[0]) / h,
    "backward_en_t4": (velocidades[4] - velocidades[3]) / h,
    "centered_en_t2": (velocidades[3] - velocidades[1]) / (2 * h),
}

# 2) Presion estelar
h = 0.1
presiones = np.array([10.00, 9.15, 8.20, 7.12, 5.90])
respuesta_2 = (3 * presiones[4] - 4 * presiones[3] + presiones[2]) / (2 * h)

# 3) Redshift
h = 2.0
z = np.array([0.0020, 0.0035, 0.0055, 0.0080, 0.0110])
respuesta_3a = (z[3] - z[1]) / (2 * h)
respuesta_3b = (z[3] - 2 * z[2] + z[1]) / (h**2)

print("Respuesta 1:", respuesta_1)
print("Respuesta 2:", respuesta_2)
print("Respuesta 3a:", respuesta_3a)
print("Respuesta 3b:", respuesta_3b)
```

## Resumen de entrega

- `forward en t0 = 10.4 km s^-2`
- `backward en t4 = 13.2 km s^-2`
- `centered en t2 = 11.7 km s^-2`
- `dP/dr = -12.9`
- `dz/dD = 0.001125 Mpc^-1`
- `d^2z/dD^2 = 0.000125 Mpc^-2`
